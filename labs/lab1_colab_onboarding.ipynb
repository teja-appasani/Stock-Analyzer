## Section 5 — A tiny preview of pandas

We'll learn pandas properly in **Week 4**, but here's a 30-second preview of what it does.

A **DataFrame** is like a spreadsheet inside Python — rows, columns, you can filter, group, and plot.

````python
import pandas as pd

# Build a tiny DataFrame
df = pd.DataFrame({
    "song": ["Mr. Brightside", "Pink + White", "Motion Sickness"],
    "artist": ["The Killers", "Frank Ocean", "Phoebe Bridgers"],
    "duration_s": [223, 203, 243]
})

df
````

## Important details and common questions

- **Drive, GitHub, and Gradescope do different jobs.** Drive is your working copy, GitHub is an optional version-controlled backup, and the downloaded notebook uploaded to Gradescope is the official submission.
- **Colab state is temporary.** Variables and uploaded files disappear when the runtime resets. A notebook is ready only when `Runtime -> Restart session and run all` succeeds from the top.
- **Do not place passwords, tokens, or private data in a notebook or repository.** A private repository reduces visibility; it is not a substitute for handling secrets correctly.
- **Execution order matters.** A later cell can use a name only after the earlier cell defining it has run in the current runtime.
